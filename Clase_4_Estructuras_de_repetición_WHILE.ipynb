Crea un ejercicio que resuelva una ecuación cuadrática pidiendo los valore de a, b y c

````python
print("Recordando que una ecuación cuadrática es de la forma ax**2+bx+c")
a=int(input("Ingrese el valor de a: "))
b=int(input("Ingrese el valor de b: "))
c=int(input("Ingrese el valor de c: "))
````

````python
x1=-b+(b**2-4*a*c)**1/2/(2*a)
x2=-b-(b**2-4*a*c)**1/2/(2*a)
print(f"Los resultados serían x={x1} y x={x2}")
````

#ESTRUCTURA WHILE - MIENTRAS
No se sabe cuántas veces se debe repetir la acción

Sintaxis en python
```
while (CI):
  parte del código a repetir
  Modificador de CI
```
**OJO!!**
CI: La CI debe tener un valor de verdad VERDADERO (TRUE) para poder ingresar al while.



Sintaxis en java
```
while(CI){
  Código que debe repetirse
  Un modificador de CI
}
```
