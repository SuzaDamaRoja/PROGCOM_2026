```python
flores={
    "tipo":["margarita","azucena","rosa"],
    "cant":[10,20,30],
    "precios":[1200,3000,1200]
}
```

```python
def menu():
  """
  Llama el tipo de flores en una floristería en forma de tabla
  """
  print("Se ofrecen los siguientes productos: \n")

  # Encabezados de la tabla con alineación
  print(f"{'Flores':<15} | {'Cantidad':<10} | {'Precio Unitario':<15}")
  print("-" * 46)

  # Recorremos las tres listas simultáneamente usando zip
  for tipo, cantidad, precio in zip(flores['tipo'], flores['cant'], flores['precios']):
    print(f"{tipo.capitalize():<15} | {cantidad:<10} | ${precio:<14}")
```

```python
def seleccion():
  """
  Imprime el menú de opciones
  Retorno:
  opc: str
  Es una variable que indica la opción del usuario
  """
  print("\nQué desea hacer?")
  print(f"{1}. Comprar flores")
  print(f"{2}. Pagar")
  print(f"{3}. Generar factura")
  print(f"{4}. Salir")
  opc = input("Seleccione su opción: \n")
  return opc
```

```python
def compra_flores(flores,compra_user):
          #Falta colocar stock
          #Qué pasa si el usuario entra y ya no hay flores
          #Quitar del carrito
            print("Ha seleccionado: Comprar flores\n")
            flor_opc = input("Seleccione la flor que desea comprar: ").lower().strip()
            if flor_opc in flores['tipo']:
                idx = flores['tipo'].index(flor_opc)
                flor_cant = int(input("Ingrese la cantidad que desea comprar: "))

                if flor_cant <= flores['cant'][idx]:
                    # Agregamos los valores a las listas del carrito
                    compra_user['tipo'].append(flor_opc)
                    compra_user['cantidad'].append(flor_cant)
                    compra_user['precio'].append(flores['precios'][idx])

                    # Restamos del inventario
                    flores['cant'][idx] -= flor_cant
                    print(f"¡Se han añadido {flor_cant} {flor_opc}(s) al carrito!")
                    # Removemos del inventario las flores si llegan a 0 usando .pop()
                    if flores['cant'][idx] == 0:
                        flores["tipo"].pop(idx)
                        flores["cant"].pop(idx)
                        flores["precios"].pop(idx)

                else:
                    print("Cantidad no disponible")
            else:
                print(f"La flor {flor_opc} no está disponible")
            return flores,compra_user
```

```python
def pagar(compra_user):
  """
  Qué hace la función
  Parámetros:
  compra_user: dict
  compra_user={
              "tipo":[],
              "cantidad":[],
              "precio":[]
  }
  """
            #Tabla de compras del usuario
            if len(compra_user["tipo"]) > 0:
                print("Seleccione el medio de pago: \n")
                print(f"{1}. Efectivo")
                print(f"{2}. Tarjeta de crédito")
                print(f"{3}. Tarjeta de débito")
                print(f"{4}. Otro medio de pago")
                pago_opc = input("Seleccione su opción: \n")

                # Calculamos el total multiplicando cada cantidad por su precio unitario
                total_pagar = 0
                for i in range(len(compra_user["tipo"])):
                    subtotal = compra_user["cantidad"][i] * compra_user["precio"][i]
                    total_pagar += subtotal
                #Si se compra por efectivo o débito--> si hay plata
                #Si se compra con tarjeta --> cuotas

                print(f"\nUsted ha seleccionado la opción de pago {pago_opc}.")
                print(f"El total a pagar por su compra es: ${total_pagar}")

            else:
                print("El carrito de compras está vacío. ")
            return total_pagar
```

```python
def factura(compra_user,total_pagar):
  print("Ha seleccionado: Generar factura\n")
  # Encabezados de la tabla con alineación
  print(f"{'Flores':<15} | {'Cantidad':<10} | {'Precio Unitario':<15} | {'Subtotal':<15}")
  print("-" * 46)
  for tipo, cantidad, precio in zip(compra_user['tipo'], compra_user['cantidad'], compra_user['precio']):
     print(f"{tipo.capitalize():<15} | {cantidad:<10} | ${precio:<14} | ${precio*cantidad:<14} ")
```

```python
print("Bienvenido a mi floristería")
# Inicializamos el carrito de compras fuera del ciclo para que no se borre al repetir el menú
flores={
    "tipo":["margarita","azucena","rosa"],
    "cant":[10,20,30],
    "precios":[1200,3000,1200]
}
compra_user = {
    'tipo': [],
    'cantidad': [],
    'precio': []
}

while True:
    opc=seleccion()
    match opc:
        case "1":
          menu()
          compra_flores(flores,compra_user)
        case "2": total_pagar=pagar(compra_user)
        case "3":
            factura(compra_user,total_pagar)
            break
        case "4":
            print("Ha seleccionado: Salir\n")
            break
        case _:
            print('Número incorrecto, por favor seleccione uno correcto')

print("Gracias por emplear nuestros servicios. Feliz día")
```
