#WHILE
1. Cuando no se sabe cuántas veces se repite una acción
2. Es gobernado por una condición de parada
3. Es necesaria una condición inicia, la cuál **DEBE** cambiar adentro del ciclo.

Sintaxis
```
while (condición):
  código a repetir
  modificación de condición
```
**OJO**:
La modificación de la condición puede darse de diferentes forma:
1. Contador
2. Un acumulador
3. Cambio de bandera (trabajando con booleanos)
4. Con un input.

````python
comida=["combo 1","combo 2","combo 3", "combo 4", "papas", "jugos naturales", "cerveza", "agua", "gaseosas"]
precio=[20000,25000,18000,24500,3000,7000,3500,4000,2500]

venta_dia = 0

print("Recaudación diaria")

for i  in range(len(comida)):
  cantidad = int(input(f"¿Cuantas unidades de {comida[i]} Se vendieron?"))
  subtotal = precio[i]*cantidad
  venta_dia = venta_dia + subtotal

print(f"El total recaudado en el dia es: ${venta_dia}")
````

````python
comida=["combo 1", "combo 2", "combo 3", "combo 4", "papas", "jugos naturales", "cerveza", "agua", "gaseosas"]
precios=[20000, 25000, 18000, 24500, 3000, 7000, 3500, 4000, 2500]
cuenta=0
pedido="s"
while pedido=="s":
  ans=input("¿Qué desea ordenar? ").lower()
  if ans in comida:
    cantidad=int(input("¿Cuántas unidades quiere? "))
    cuenta+=precios[comida.index(ans)]*cantidad
  else:
    pedido=input("Hay otro cliente? s/n= ")
print("El total de dinero de las ventas del día fue: ",cuenta)
````

#FUNCIONES
Es un pedazo de código "muerto" que no se ejecuta hasta que el código principal lo ejecute.

Ventanas
1. Organiza el código
2. Habilita un nuevo paradigma de programación: Programación modular
3. Habilita el copiado dinámico (reutilización del código)

Sintaxis
```
def nombre_funcion(p1,p2,p3,...,pn):
  aquí va el código
  return (dato)
```
Donde:
* p1, p2, .... pn son parámetros
* return retorna un dato

Existen 4 tipos diferentes de funciones
1. Sin parámetro, sin retorno
2. Sin parámetro, con retorno
3. Con parámetro, sin retorno
4. Con parámetro, con retorno
