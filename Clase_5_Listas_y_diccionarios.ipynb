````python
#Quiero saber cuántos de n estudiantes sacan más que el promedio
suma=0
c=0
n=int(input("Ingrese la cantidad de estudiantes"))
for i in range(n):
  nota=float(input(f"Ingrese la nota del estudiante {i}:"))
  suma+=nota
prom=suma/n
for i in range(n):
  nota=float(input(f"Ingrese la nota del estudiante {i}:"))
  if nota>prom:
    c=c+1
print(f"{c} estudiantes sacaron mayor nota que el promedio.")
````

#LISTAS
Es una colección de elementos que cumplen las siguientes características.
1. Mutable
2. Es que tiene index
3. Sus elementos están separados por comas

##Creación de listas
nombre_lista=[]

Existen dos formas de crear una lista:

2. Por comprensión: Se enuncian las características comunes de los elementos

##RECORRIENDO LAS LISTAS DE FORMA AUTOMÁTICA



````python
for caja in range(3):
  print(f"{Galleta[caja]}:{cant_galletas[caja]}")
````

````python
suma=0
c=0
notas=[]
n=int(input("Ingrese la cantidad de estudiantes"))
for i in range(n):
  nota=float(input(f"Ingrese la nota del estudiante {i}:"))
  notas.append(nota)
  suma+=nota
prom=suma/n
for i in range(n):
  if notas[i]>prom:
    c=c+1
print(f"{c} estudiantes sacaron mayor nota que el promedio.")
````

###2. Verificar si existe un elemento en la lista

````python
user=input("Ingrese la galleta que quiere comprar")
for galleta in Galleta:
  if galleta==user:
    print("Si hay 🍪")
    break
  else: print("No hay😿")
````

````python
for galleta in Galleta:
  if galleta==user:
    print("Si hay 🍪")
    break
else: print("No hay😿")
````

````python
for galleta in Galleta:
  if user==Galleta[0]:
    print("Si hay 🍪")
    break
  elif user==Galleta[1]:
    print("Si hay 🍪")
    break
  elif user==Galleta[2]:
    print("Si hay 🍪")
    break
  else:
    print("No hay😿")
    break
````
